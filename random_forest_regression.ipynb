```python
import pandas as pd
```

```python
read_data=pd.read_csv("50_Startups.csv")
```

```python
read_data.head()
```

```python
data_set=pd.get_dummies(read_data,drop_first=True).astype(int)
```

```python
data_set.head()
```

```python
data_set.columns
```

```python
# distribute feature and target:
inputs=data_set[['R&D Spend', 'Administration', 'Marketing Spend','State_Florida', 'State_New York']]
outputs=data_set['Profit']
```

```python
# split:
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(inputs,outputs,test_size=0.3,random_state=42)
```

```python
# learning-random forest:
from sklearn.ensemble import RandomForestRegressor
rd=RandomForestRegressor(n_estimators=80,random_state=42)
rf=rd.fit(x_train,y_train)
```

```python
# evaluation:
y_pred=rf.predict(x_test)
```

```python
from sklearn.metrics import r2_score
r2=r2_score(y_test,y_pred)
```
